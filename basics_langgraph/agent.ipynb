```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
```

```python
@tool
def add(a: float, b: float) -> float:
    """Add two numbers together and return the sum."""
    return a + b

@tool
def multiple(a: float, b: float) -> float:
    """Multiple two numbers together and return the sum."""
    return a * b
```

```python
llm=ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
)

tools=[add,multiple]
agent=create_agent(llm,tools=tools)
```

```python
from langchain.messages import HumanMessage

result=agent.invoke({"messages":[HumanMessage(content="What is (6 * 7) + 15?")]})
```

```python
print("=" * 60)

for msg in result["messages"]:
    role = msg.__class__.__name__

    # Show tool calls if present (these are the LLM's action requests)
    if hasattr(msg, 'tool_calls') and msg.tool_calls:
        print(f"[{role}] (Thought + Action request)")

        for tc in msg.tool_calls:
            print(f"  -> tool: {tc['name']}, args: {tc['args']}")

    else:
        print(f"[{role}] {msg.content}")

print("=" * 60)
```
